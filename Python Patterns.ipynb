```python
#16.] Count the total number of digits in a given number

num = int(input("Enter a number: "))
count = 0
n = abs(num)  # handle negative numbers
while n > 0:
    n //= 10
    count += 1
print("Total digits:", count)
```

```python
#17.] Accept any number from the user and check whether it is a perfect number or not
num = int(input("Enter a number: "))
sum_div = 0
for i in range(1, num):
    if num % i == 0:
        sum_div += i
if sum_div == num:
    print(num, "is a perfect number")
else:
    print(num, "is not a perfect number")
```

```python
#18.] Write a program to calculate the sum of series upto n terms 
#for a given number. Eg. If number is 2 and n_terms=5 the series will be 
#2+22+222+2222+22222=24690
number = int(input("Enter number: "))
n_terms = int(input("Enter number of terms: "))
sum_series = 0
term = 0
for i in range(n_terms):
    term = term*10 + number
    sum_series += term
print("Sum of series:", sum_series)
```

```python
#19.] Print all perfect numbers between 1-500
for num in range(1, 501):
    sum_div = 0
    for i in range(1, num):
        if num % i == 0:
            sum_div += i
    if sum_div == num:
        print(num)
```
