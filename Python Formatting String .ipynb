```python
#4.] Write a Python program to create all possible strings by using 
#'a', 'e', 'i', 'o', 'u'. Use the characters exactly once. 
import itertools

chars = ['a', 'e', 'i', 'o', 'u']
perms = list(itertools.permutations(chars))

for p in perms:
    s = ''.join(p)
    print(f"{s}")

```

```python
#5.] Write a Python program to print the number of prime numbers 
#which are less than or equal to an given integer. 
#Input: Input the number(n):35
#Number of prime numbers which are less than or equal to n.: 11
n = int(input("Input the number (n): "))
count = 0

for num in range(2, n+1):
    for i in range(2, int(num**0.5)+1):
        if num % i == 0:
            break
    else:
        count += 1

print(f"Number of prime numbers which are less than or equal to {n}: {count}")
```

```python
#7.] Write a Python program to find the longest common prefix 
#string amongst an given array of strings. Return false If there is no common prefix.
#For Example, longest common prefix of "abcdefgh" and "abcefgh" is "abc". 
#Sample Output : abc 
def longest_common_prefix(strings):
    if not strings:
        return False
    prefix = strings[0]
    for s in strings[1:]:
        while not s.startswith(prefix):
            prefix = prefix[:-1]
            if prefix == "":
                return False
    return prefix

arr = ["abcdefgh", "abcefgh"]
result = longest_common_prefix(arr)

print(f"Longest common prefix: {result}" if result else "No common prefix found")
```

```python
#14.] Write a Python program to find the first appearance of the 
#substring 'not' and 'poor' from a given string, if 'not' follows the 
#'poor', replace the whole 'not'...'poor' substring with 'good'. 
#Return the resulting string.   
#Sample String : 'The lyrics is not that poor!'
#'The lyrics is poor!'
#Expected Result : 'The lyrics is good!'
#'The lyrics is poor!'
s = input("Enter a string: ")

not_pos = s.find("not")
poor_pos = s.find("poor")

if not_pos != -1 and poor_pos != -1 and not_pos < poor_pos:
    s = f"{s[:not_pos]}good{s[poor_pos+4:]}"

print(f"Result: {s}")
```

```python
#18.] Given a string input Count all lower case, upper case, digits, 
#and special symbols
#eg :    Total counts of chars, digits,and symbols Chars = 8 Digits = 3 
#Symbol = 4
s = input("Enter a string: ")

lower = upper = digits = symbols = 0

for c in s:
    if c.islower():
        lower += 1
    elif c.isupper():
        upper += 1
    elif c.isdigit():
        digits += 1
    else:
        symbols += 1

print(f"Lowercase letters: {lower}")
print(f"Uppercase letters: {upper}")
print(f"Digits: {digits}")
print(f"Special symbols: {symbols}")
```

```python
#19.] Given two strings, s1 and s2, create a mixed String
#Note: create a third-string made of the first char of the last char of b, 
#the second char of the second last char of b, and so on. 
#Any leftover chars go at the end of the result.
#eg:  s1 = "Pynative"
     #s2 = "Website"
#o/p s3 = PeytniastbievWe

s1 = input("Enter s1: ")
s2 = input("Enter s2: ")

# Mix s1 and reversed s2
length = min(len(s1), len(s2))
s3 = ''.join([s1[i] + s2[-(i+1)] for i in range(length)])
# Add leftover characters
if len(s1) > len(s2):
    s3 += s1[length:]
elif len(s2) > len(s1):
    s3 += s2[:len(s2)-length]

print(f"Mixed string: {s3}")
```

```python
#20.] String characters balance Test. We ll say that a String s1 and s2 is 
#balanced if all the chars in the string1 are there in s2. characters 
#position doesn t matter.
#eg : For Example:   (yn, Pynative) = True

s1 = input("Enter s1: ")
s2 = input("Enter s2: ")

balanced = all(c in s2 for c in s1)
print(f"Is balanced: {balanced}")
```

```python
#22.] Given a string, return the sum and average of the digits that appear 
#in the string, ignoring all other characters
#eg :     input = "English = 78 Science = 83 Math = 68 History = 65"
#output =  sum 294 Percentage is 73.5

s = "English = 78 Science = 83 Math = 68 History = 65"
digits = [int(c) for c in s if c.isdigit()]
total = sum(digits)
average = total / len(digits)
print(f"Sum: {total}  Average: {average}")
```

```python
#23.] Given an input string, count occurrences of all characters within a string
s = input("Enter a string: ")
counts = {}
for c in s:
    counts[c] = counts.get(c, 0) + 1

for char, count in counts.items():
    print(f"'{char}': {count}")
```
